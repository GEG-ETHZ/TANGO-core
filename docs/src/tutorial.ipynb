# Tutorial

In this tutorial we demonstrate the use of TANGO in a simple example. 

## Defining the workflow

The workflow is defined in a YAML file as a graph where the nodes represent the
different components (processsing units) of the system and the edges represent
the connections between the components. The workflow is defined in the `graph`
section of the YAML file.

```yaml
graph:
  nodes:
    inj_1: # root node
      type: 'injection_well'
      unit: 'native'
    reservoir:
      type: 'reservoir'
      unit: 'native'
    prd_1:
      type: 'production_well'
      unit: 'native'
    surface:
      type: 'surface'
      unit: 'native_CPG'
  edges:
    1:
      source: 'surface'
      target: 'inj_1'
    2:
      source: 'inj_1'
      target: 'reservoir'
    3:
      source: 'reservoir'
      target: 'prd_1'
    4:
      source: 'prd_1'
      target: 'surface'

# simulation input

runspec:
  title: 'CPG DOUBLET'
  workflow: 'native'

props:
  fluid_name: 'CO2'
  brine_density: 1000 # kg/m3
  rock_density: 2650.0 # kg/m3
  rock_cp: 1000.0  # # specific heat capacity (J/(kg*K))
  rock_k: 2.1  # thermal conductivity (W/(m*K))
  thermal_gradient: 0.035  # degC/m

surface:
  surface_temperature: 288.15  # K
  surface_pressure: 100_000  # Pa
  pipe_diameter: 0.41 # m
  approach_temperature: 7.0  # K
  condensation_pressure_offset: 50_000.0  # Pa
  eta_turbine: 0.78  # (-)

reservoir:
  reservoir_depth: 2500.0 # m
  thickness: 300.0 # m
  permeability: 5.0e-14    # m2 ~50mD
  porosity: 0.1

injection_well:
  well_length: 2500. # m
  diameter: 0.410 # m

production_well:
  well_length: 2500. # m
  diameter: 0.410 # m

schedule:
  mass_flow_rate: 200.0  # kg/s (injection rate per pattern)
  well_pattern: 'DOUBLET' # doublet, inverted_5spot, 5spot
  well_spacing: 707. # m (inj-prod pair)
  well_pattern_dimens: [1,1] # [num_patterns_x_dir, num_patterns_y_dir]
  start_date: '2019-01-01' # Year-mm-dd
  num_tsteps: 5 # number of time steps
  dt: 1.0  # time step increment (year)

economic: # optional
  type: 'native'
  project_maturity: 'green_field' # brown_field or green_field
  discount_rate: 0.095
```
